```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

root = Path("../data/raw/asl_alphabet_train")

# each subfolder is one class
classes= sorted([p.name for p in root.iterdir() if p.is_dir()])
print(f"Number of classes: {len(classes)}")

# display all the sub folders and number of files in this folder
for c in classes:
    folder= root/c
    num_files= len(list(folder.glob("*")))
    print(f"{c}: {num_files} files")
    #run and check the sizes and modes of the images in this folder
    sizes= []
    modes= []
    for f in folder.glob("*"):
        with Image.open(f) as img:
            sizes.append(img.size)
            modes.append(img.mode)
    print(f"Sizes: {set(sizes)}")
    print(f"Modes: {set(modes)}")
```
